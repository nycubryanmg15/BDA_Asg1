```python
import torch
from transformers import AutoModelForCausalLM
import matplotlib.pyplot as plt

model = AutoModelForCausalLM.from_pretrained('meta-llama/Llama-3.2-1B',)

print('done loading')
```

```python
for k in [0,7,15]:
    q_weight = model.model.layers[k].self_attn.q_proj.weight.detach()
    
    U, S, Vh = torch.linalg.svd(q_weight.float(), full_matrices=False)
    
    energy_ratios = torch.cumsum(S ** 2, dim=0) / torch.sum(S ** 2) * 100

    rank_50pct = int((energy_ratios < 50).sum().item()) + 1


    ### i. Cumulative energy (%) using a line chart ###
    # plt.figure(figsize=(7, 4))
    plt.plot(range(1,201), energy_ratios[:200].detach().cpu().numpy())
    plt.xlabel("Target rank")
    plt.ylabel("Cumulative Energy (%)")
    plt.title(f"Layer {k}: Cumulative energy, Min.Rank = {rank_50pct}")
    plt.ylim(0, 100)
    plt.grid(True, alpha=0.3)
    plt.show()
    ### i. end ###

    ### ii. Singular value magnitudes using a bar chart ###
    # plt.figure(figsize=(7, 4))
    plt.bar(range(1,201),S[:200])
    plt.xlabel("Singular Value Index")
    plt.ylabel("Singular Value Magnitude")
    plt.title(f"Layer {k}: Singular Value Magnitude")
    plt.grid(True, alpha=0.3)
    plt.show()
    ### ii. end ###





```
